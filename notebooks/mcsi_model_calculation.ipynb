## 🛠️ Module 2: Data Hygiene & Algorithmic Feature Engineering

This cell executes automated data cleaning protocols to standardize the column string layout before computing dynamic risk metrics. It systematically strips out trailing blank space anomalies and casts headers to absolute lowercase text.

Once normalized, the data engine computes a new calculated field—**`composite_risk_score`**—by dynamically multiplying the baseline EWG hazard values against the custom public health weights established within `RISK_GUIDE.md`.

## 🧮 Module 3: Advanced Categorical Aggregation & Class Sorting

This module transitions the repository into Exploratory Data Analysis (EDA) by executing relational aggregations using `.groupby()` and `.agg()`. The script segments the chaotic 30-row master dataset into its five core functional chemical classes, automatically computing the compound count distributions and mathematical risk averages to identify the most critical exposure vectors.


```python
# 1. AUTOMATED DATA CLEANING: Strip out hidden empty spaces and force lowercase headers
df.columns = df.columns.str.strip().str.lower()

# 2. FEATURE ENGINEERING: Compute the customized Composite Risk Score cell array
df['composite_risk_score'] = df['toxicity_score_ewg'] * df['risk_weight_multiplier']

# 3. DATA PREVIEW: View newly engineered master data table structure
print("=== 🚀 DATA CLEANING & FEATURE ENGINEERING COMPLETE ===")
df[['chemical_id', 'chemical_name', 'composite_risk_score']].head()
```

## 📈 Module 4: Statistical Visualizations & Pearson Correlation Matrix

To mathematically validate the structural integrity of the MCSI multi-tiered scoring methodology, this module utilizes **Matplotlib** and **Seaborn** to plot a Pearson correlation matrix heatmap. This visual representation maps out the linear relationship coefficients between your raw toxicity scores, risk multipliers, and final engineered values to verify total algorithmic symmetry.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. CORE MATH: Compute the Pearson correlation matrix over your numerical fields
correlation_matrix = df[['toxicity_score_ewg', 'risk_weight_multiplier', 'composite_risk_score']].corr(method='pearson')

# 2. VISUAL CHARTING: Generate a color-coded visual heatmap window with decimal annotations
plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)

# 3. INTERFACE DISPLAY: Add technical title headers and render the chart graphic layout
plt.title("=== 📈 MCSI TOXICOLOGICAL CORRELATION HEATMAP ===")
plt.show()
```

**Data Insight:** Linear correlation modeling establishes an elite Pearson coefficient of 0.85 between raw environmental hazards (`toxicity_score_ewg`) and structural multicultural parameters (`risk_weight_multiplier`).
